# Exploring multi-agent communication and actions

This notebook reads the **extracted** episode JSON. The pilot diners run establishes pipeline function, **not collusion detection**.

```python
from pathlib import Path
import json

path = Path("../results/diners_smoke_episodes.json")
if not path.exists():
    print("Run python -m src.pipeline from the repository root first.")
    episodes = []
else:
    episodes = json.loads(path.read_text())
print("Episodes:", len(episodes))
```

```python
for episode in episodes[:3]:
    print("Episode:", episode["episode_id"])
    print("Messages:", episode["communication"])
    print("Actions:", episode["actions"])
    print("Collusion ground truth:", episode["collusion_label"])
```
