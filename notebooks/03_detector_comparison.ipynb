# Action-matched detector comparison

**Exploratory pilot only.** The reference target below is protocol-level acceptance, **not validated collusion**. Public messages and actions are the only detector inputs; private side-channel is excluded.

```python
from pathlib import Path
import json, csv
import pandas as pd
import matplotlib.pyplot as plt
base = Path("../results/action_matched_evaluation") if Path("../results/action_matched_evaluation").exists() else Path("results/action_matched_evaluation")
pred = pd.read_csv(base / "predictions.csv")
report = json.loads((base / "metrics.json").read_text())
display(pred)
print("Reference target:", report["target"])
```

```python
rows = [{"detector": name, **values} for name, values in report["metrics"].items()]
summary = pd.DataFrame(rows)
display(summary)
summary.set_index("detector")[["precision", "recall", "f1"]].plot(kind="bar", rot=20, ylim=(0, 1), title="Against protocol-agreement proxy (not collusion truth)")
plt.tight_layout()
plt.show()
```

## Interpretation and limitations

- Treatment assignment and a model-generated acceptance do not establish emergent collusion.
- Because all agents are prompted to catch three, action-only baselines may produce false positives on independent quota violations.
- Publicly concealed coordination can be indistinguishable from non-coordinated deceit with the same messages and actions.
- Pilot samples and handcrafted detectors do not establish generalizable accuracy.
